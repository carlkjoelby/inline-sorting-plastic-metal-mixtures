```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

df2 = pd.read_csv("C:/Users/chris/Desktop/Bachelor/Calc/Plot/CSV/20mmR20.csv")
df2.columns = ['x_350', 'v_350', 'x_150', 'v_150']

df3 = pd.read_csv("C:/Users/chris/Desktop/Bachelor/Calc/Plot/CSV/30mmR20.csv")
df3.columns = ['x_350', 'v_350', 'x_150', 'v_150']

df4 = pd.read_csv("C:/Users/chris/Desktop/Bachelor/Calc/Plot/CSV/40mmR20.csv")
df4.columns = ['x_350', 'v_350']

avg_v350 = df2['v_350'].mean()
print(f"Avg 40mm (3.0): {avg_v350:.3f} m/s")

plt.figure(figsize=(12, 5))

plt.scatter(df2['x_350'], df2['v_350'], label='20mm', color='tab:blue')
plt.scatter(df3['x_350'], df3['v_350'], label='30mm', color='tab:orange')
plt.scatter(df4['x_350'], df4['v_350'], label='40mm', color='tab:green')

plt.ylim(0.39, 0.45)
plt.xlabel('Position (m)', fontsize=14)
plt.ylabel('Hastighed (m/s)', fontsize=14)
plt.title('Prototype 3.0\nHastighedsprofiler for forskellige dyse længder', fontsize=16)

plt.tick_params(axis='both', labelsize=12)

plt.grid(True, which='major', alpha=0.6)

plt.axvline(x=-0.065, color='black', linestyle='-', linewidth=2, alpha=0.8, label='Væg')
plt.axvline(x=0.065,  color='black', linestyle='-', linewidth=2, alpha=0.8)
plt.axhline(y=avg_v350, color='red', linestyle='--', linewidth=1.5, alpha=0.8,
            label=f'Gennemsnits hastighed: {avg_v350:.3f} m/s')

extra_text = Line2D(
    [], [],
    linestyle='none',
    label=f'Antal huller: 26,   L/min: 65'
)

handles, labels = plt.gca().get_legend_handles_labels()
handles.append(extra_text)

plt.legend(
    handles=handles,
    fancybox=True,
    framealpha=1,
    facecolor="white",
    edgecolor="black",
    fontsize=12
)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

df1 = pd.read_csv("C:/Users/chris/Desktop/Bachelor/Calc/Plot/CSV/Før.txt", skiprows=1, header=None,
                  names=['distance', 'x_coord', 'y_coord', 'u_ms'])
df1 = df1.dropna(subset=['u_ms'])

df2 = pd.read_csv("C:/Users/chris/Desktop/Bachelor/Calc/Plot/CSV/Efter.txt", skiprows=1, header=None,
                  names=['distance', 'x_coord', 'y_coord', 'u_ms'])
df2 = df2.dropna(subset=['u_ms'])

df1['y_centered'] = df1['y_coord'] - df1['y_coord'].mean()
df2['y_centered'] = df2['y_coord'] - df2['y_coord'].mean()

avg_v1 = df1['u_ms'].mean()
avg_v2 = df2['u_ms'].mean()
print(f"Datasæt 1 gennemsnit: {avg_v1:.3f} m/s")
print(f"Datasæt 2 gennemsnit: {avg_v2:.3f} m/s")

plt.figure(figsize=(12, 5))

plt.scatter(df1['y_centered'], df1['u_ms'], color='steelblue')
plt.scatter(df2['y_centered'], df2['u_ms'], color='darkorange')

plt.xlabel('Position (m)', fontsize=14)
plt.ylabel('Hastighed (m/s)', fontsize=14)
plt.title('Målt hastighedsprofil for prototype 3.1', fontsize=16)

plt.tick_params(axis='both', labelsize=12)

plt.grid(True, which='major', alpha=0.6)

plt.axvline(x=df1['y_centered'].min(), color='black', linestyle='-', linewidth=2, alpha=0.8, label='Væg')
plt.axvline(x=df1['y_centered'].max(), color='black', linestyle='-', linewidth=2, alpha=0.8)
plt.axhline(y=avg_v1, color='steelblue', linestyle='--', linewidth=1.5, alpha=0.8)
plt.axhline(y=avg_v2, color='darkorange', linestyle='--', linewidth=1.5, alpha=0.8)

# Samlede legend entries med scatter + gennemsnit på én linje
legend_før = Line2D([0], [0], marker='o', color='steelblue', linestyle='--',
                    label=f'Før (gns: {avg_v1:.3f} m/s)', markersize=6)
legend_efter = Line2D([0], [0], marker='o', color='darkorange', linestyle='--',
                      label=f'Efter (gns: {avg_v2:.3f} m/s)', markersize=6)
legend_vaeg = Line2D([0], [0], color='black', linestyle='-', linewidth=2, label='Væg')
extra_text = Line2D([], [], linestyle='none',
                    label='350mm over dysen\nLængde af huller: 40mm (Ø3mm)\nAntal huller: 26,  L/min 20')

plt.legend(
    handles=[legend_før, legend_efter, legend_vaeg, extra_text],
    fancybox=True,
    framealpha=1,
    facecolor="white",
    edgecolor="black",
    fontsize=12,
    loc='upper left'
)

plt.show()
```
